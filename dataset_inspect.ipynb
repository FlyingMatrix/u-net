```python
import numpy as np
import matplotlib.pyplot as plt
import random
from IPython.display import display
from train import data_prep
```

```python
# data preparation
train_loader, val_loader = data_prep()

# get the first batch
train_iter = iter(train_loader)
images, masks = next(train_iter)

# print batch data shape
print(">>> Images shape of the first batch: " + str(images.shape))
print(">>> Masks shape of the first batch: " + str(masks.shape))
```

```python
# get a random sample from the first batch
idx = random.randint(0, 4)
image = images[idx] # torch.Size([1, 512, 512])
mask = masks[idx] # torch.Size([512, 512])

# visualization of the sample data
fig, axes = plt.subplots(1, 2, figsize=(8, 4))

axes[0].imshow(image.squeeze().cpu().numpy(), cmap="gray")
axes[0].axis("off")
axes[0].set_title("Image")

axes[1].imshow(mask.squeeze().cpu().numpy(), cmap="gray")
axes[1].axis("off")
axes[1].set_title("Mask")

plt.tight_layout()
plt.show()
```

```python
# print the image matrix
np.set_printoptions(threshold=np.inf)
print(">>> Image matrix: ")
display(image.squeeze().cpu().numpy()) # display(): show rich content inline
print(">>> Image shape: " + str(image.squeeze().cpu().numpy().shape))
```

```python
# print the mask matrix
print(">>> Mask matrix: ")
display(mask.squeeze().cpu().numpy()) 
print(">>> Mask shape: " + str(mask.squeeze().cpu().numpy().shape))
```
